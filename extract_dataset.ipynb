```python
import os
import numpy as np
import matplotlib as plt
import pandas as pd
import torch
```

```python
with open('poems_dataset.csv', 'r', encoding='utf-8') as f:
    csvFile = pd.read_csv(f)

dataset = csvFile.drop(columns=['id', 'url']).dropna(subset=['title'])
dataset.info()
dataset.head()
```

```python
dataset.convert_dtypes(infer_objects=False)
dataset_7chu = dataset[dataset['genre'] == '7 chu'].drop(columns=['genre'])
dataset_7chu.info()
dataset_7chu.head()
```

```python
flattened_dataset_7chu = dataset_7chu[['title', 'content']].values.flatten()
with open('dataset_7chu.txt', 'w', encoding='utf-8') as file:
    index = 0 
    for value in flattened_dataset_7chu:
        file.write(f"{value}")
        if index % 2 == 0:
            file.write(':\n')
        else:
            file.write('\n\n')
        index += 1
```
